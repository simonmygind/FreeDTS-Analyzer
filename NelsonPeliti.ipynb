```python
N_r = 15

Simulation = "OnlyMembrane"
EPS = "no"
AnalysisFile=f"{FrameTension}/{InitialConfiguration}/{Simulation}/VFF{VFF}/EPS{EPS}/StaticUndulationSpectrum/{FTMethod}{N_k}.txt"
DataFile=f"/lustre/astro/semygind/Data/MembraneInclusion/{FrameTension}/{InitialConfiguration}/{Simulation}/VFF{VFF}/EPS{EPS}/"
data = load_boxsize_fourier_file(AnalysisFile)
q_arr, hqn_arr = data[('q','<u u>')]      # each (N_unique, N_r)
q = q_arr.mean(axis=1)                    # mean q per bin (box size varies slightly per replica)

# Bare membrane
Lx, Ly, Lz = calculate_L_replica(DataFile, N_r=N_r)   # each (N_r,), per-replica box size

kappa_eff_r = N_v * k_bT / (Lx[None,:] * Ly[None,:] * hqn_arr * q_arr**4)  # (N_unique, N_r)
kappa_eff = kappa_eff_r.mean(axis=1)
kappa_eff_err = kappa_eff_r.std(axis=1, ddof=1) / np.sqrt(N_r)  # SEM across replicas

fig, ax = plt.subplots()
ax.errorbar(q, kappa_eff, yerr=kappa_eff_err, fmt='.', label='Bare membrane')

# Vector field simulations
EPS_list = ["1","2","5"]
Simulation="OnlyVectorField"
for EPS in EPS_list:
    AnalysisFile=f"{FrameTension}/{InitialConfiguration}/{Simulation}/VFF{VFF}/EPS{EPS}/StaticUndulationSpectrum/{FTMethod}{N_k}.txt"
    DataFile=f"/lustre/astro/semygind/Data/MembraneInclusion/{FrameTension}/{InitialConfiguration}/{Simulation}/VFF{VFF}/EPS{EPS}/"
    data = load_boxsize_fourier_file(AnalysisFile)
    q_arr, hqn_arr = data[('q','<u u>')]
    q = q_arr.mean(axis=1)

    Lx, Ly, Lz = calculate_L_replica(DataFile, N_r=N_r)

    kappa_eff_r = N_v * k_bT / (Lx[None,:] * Ly[None,:] * hqn_arr * q_arr**4)
    kappa_eff = kappa_eff_r.mean(axis=1)
    kappa_eff_err = kappa_eff_r.std(axis=1, ddof=1) / np.sqrt(N_r)

    ax.errorbar(q, kappa_eff, yerr=kappa_eff_err, fmt='.', label=rf'$\epsilon={EPS}$')

ax.set_xscale('log')
#ax.set_yscale('log')
ax.set_xlabel(r'$q\,[1/l_{dts}]$')
ax.set_ylabel(r'$\kappa_{\rm eff}(q)$')
ax.legend()
```
