```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
import os

N_k=5
#KPKL = "KP5KL0"
#KPKL = "2NoAllignmentPotential"
KPKL = "1NoVectorField"
FTMethod = "LSFT"
fontsize=15
values = np.genfromtxt(
    f"/lustre/astro/semygind/Analyzer/version_1/{KPKL}/StaticUndulationSpectrum/Bootstrap/{FTMethod}{N_k}.txt",
    skip_header=7,
    delimiter=" ",
)
q   = values[:,0]
hqn = np.mean(values[:,1:],axis=1)
var  = np.var(values[:,1:],axis=1)

i1 = 0
i2 = 13
j = 2
q     = values[:,0][i1:]
hqn   = values[:,1][i1:]
var   = values[:,j][i1:]
#var   = np.ones(len(q))*q**4

def qn(X,C,n):
    return C/X**(n)

def q4(X,C):
    return C/X**(4)

methods = [q4,qn]
figsize = (11,7)
fig , ax = plt.subplots(1,2,figsize=figsize)
for i in range(2):
    par, cov = curve_fit(methods[i],q,hqn,sigma=np.sqrt(var),absolute_sigma=True)
    if (methods[i] is q4):
        ax[i].errorbar(q,hqn,yerr=np.sqrt(var)/np.sqrt(len(q)-1),fmt='.',
            label=rf"$C = {np.round(par[0],4)}\pm {np.round(np.sqrt(cov[0,0]),4)}$")
        ax[i].set_title(rf"$f(q)=Cq^{{-4}}$")
    if (methods[i] is qn):
        ax[i].errorbar(q,hqn,yerr=np.sqrt(var)/np.sqrt(len(q)-1),fmt='.',
            label=rf"$C = {np.round(par[0],4)}\pm {np.round(np.sqrt(cov[0,0]),4)}$"+"\n"+rf"$n = {np.round(par[1],2)}\pm{np.round(np.sqrt(cov[1,1]),2)}$")
        ax[i].set_title(rf"$f(q)=Cq^{{-n}}$")
        
    X = np.linspace(q[0],q[-1],1000)
    Y = methods[i](X,*par)
    ax[i].plot(X,Y,label="Fit")
    ax[i].legend()
    ax[i].set_xscale('log')
    ax[i].set_yscale('log')

fig.supxlabel(r"$q$",fontsize=fontsize)
fig.supylabel(r"$\left< h_q h^*_{q} \right>$",fontsize=fontsize)
fig.suptitle(KPKL+" using "+FTMethod+fr" with $N_k={N_k}$")
output_path = f"/lustre/astro/semygind/Analyzer/version_1/Plots/StaticFluctuation/LogLog/Nk{N_k}/"
os.makedirs(output_path,exist_ok=True)
#fig.savefig(os.path.join(output_path,f"{KPKL}{FTMethod}.pdf"))
```
